```python
import torch
import time
import pandas as pd
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForCausalLM

assert torch.cuda.is_available(), "런타임 유형을 GPU로 바꾸세요."

print("GPU 사용 가능:", torch.cuda.is_available())
print("GPU 이름:", torch.cuda.get_device_name(0))

model_id = "Qwen/Qwen3-1.7B"
tokenizer = AutoTokenizer.from_pretrained(model_id)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    device_map={"": 0},
)
model.eval()

print("모델:", model_id)
print("모델 장치:", next(model.parameters()).device)
print("모델 로딩 완료 / 추가 학습 없이 GPU 추론")
```

```python
schema = """
DuckDB tables:
master(VESSELCODE VARCHAR, DOC_NO VARCHAR)
pms(
    VESSELCODE VARCHAR,
    DOC_NO VARCHAR,
    MACH_DESC VARCHAR,
    MACH_NO VARCHAR,
    CRITICAL_YN VARCHAR
)

Rules:
master: maintenance documents.
pms: maintenance work lines.
Document key: (VESSELCODE, DOC_NO).
Join tables on BOTH VESSELCODE and DOC_NO.
주기관: MACH_DESC = 'MAIN ENGINE(DIESEL)'
중요 정비: CRITICAL_YN = 'Y'
호기: MACH_NO
Return one DuckDB SELECT statement only.
"""

questions = [
    "전체 정비 문서 수를 알려줘.",
    "선박별 정비 문서 수를 선박 코드와 함께 알려줘.",
    "주기관 정비 작업 라인 수를 알려줘.",
    "주기관 정비 작업 라인 수를 호기별로 알려줘.",
    "중요 정비 작업 라인 수를 알려줘.",
]

display(pd.DataFrame({
    "문항": [f"Q{i+1}" for i in range(len(questions))],
    "질문": questions,
}))
```

```python
records = []

for i, question in enumerate(questions, 1):
    messages = [
        {"role": "system", "content": schema},
        {"role": "user", "content": question},
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False,
    )

    inputs = tokenizer(text, return_tensors="pt").to("cuda")

    torch.cuda.synchronize()
    start = time.perf_counter()

    with torch.inference_mode():
        output = model.generate(
            **inputs,
            max_new_tokens=256,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )

    torch.cuda.synchronize()
    elapsed = time.perf_counter() - start

    sql = tokenizer.decode(
        output[0, inputs["input_ids"].shape[1]:],
        skip_special_tokens=True,
    ).strip()

    records.append({
        "id": f"Q{i}",
        "question": question,
        "generated_sql": sql,
        "generation_seconds": elapsed,
    })

    print(f"Q{i}: {question}")
    print(sql)
    print(f"생성 시간: {elapsed:.2f}초\n")

results = pd.DataFrame(records)
```

```python
display(results)

plt.figure(figsize=(8, 4))
plt.bar(
    results["id"],
    results["generation_seconds"],
    color="steelblue",
)
plt.xlabel("Question")
plt.ylabel("SQL generation time (seconds)")
plt.title("Qwen3-1.7B: Ship Maintenance Text-to-SQL on GPU")
plt.tight_layout()
plt.savefig("generation_time.png", dpi=180)
plt.show()

results.to_csv(
    "ship_sql_results.csv",
    index=False,
    encoding="utf-8-sig",
)

print("모델:", model_id)
print("GPU:", torch.cuda.get_device_name(0))
print("문항 수:", len(results))
print("측정 범위: model.generate 실행 시간")
print("첫 문항에는 초기 실행 비용이 포함될 수 있음")
print("SQL 생성 실습이며 DB 실행·정확도 평가는 포함하지 않음")

from google.colab import files
files.download("ship_sql_results.csv")
files.download("generation_time.png")
```
